<div class="alert alert-block alert-warning">
     <b>Manage Null Values</b>
</div>

```python

df = df.drop('Assembly', axis=1)
```

```python
# df.drop('country', axis=1, inplace=True)
df = df.drop('Transmission Type', axis=1)
```

```python
df.columns
```

```python

def fillnaObjectMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mode()[0])

columns = ['Company Name', 'Model Name', 'Location','Engine Type','Color','Registration Status']
fillnaObjectMode(columns)
```

```python

def fillnaIntMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mode()[0])

columns = ['Unnamed: 0','Price','Model Year','Mileage','Engine Capacity']
fillnaIntMode(columns)
```

```python
df.info()
```

<div class="alert alert-block alert-warning">
     <b>Manage Data Types</b>
</div>

```python
df['Price'] = df['Price'].astype('int64')
# df['price'].astype('int64', inplace=True)
```

```python
import numpy as np
from sklearn.preprocessing import LabelEncoder

def convertObjToInt(cols):
    for i in cols:
        data = pd.DataFrame({i: df[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(data))
        
        df[i] = data_LE.transform(df[i]).astype('int64')   # Convert to int64

columns = ['Company Name', 'Model Name', 'Location','Engine Type','Color','Body Type','Registration Status']
convertObjToInt(columns)
```

<div class="alert alert-block alert-info">
     <b>Train - Test Splitting</b>
</div>

```python
import pandas as pd
df = pd.read_csv('processed_data.csv')
# df.info()
df.columns
```

```python
X = df.drop('Price', axis=1)   
y = df['Price']                
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X,y, test_size=0.3, shuffle=False) 
```

```python
print(train_X.shape, train_y.shape)
print(test_X.shape, test_y.shape)
```

<div class="alert alert-block alert-info">
     <b>Applying Classifier / Model Training</b>
</div>

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


model = LinearRegression()


model.fit(train_X, train_y)


```

```python
predictions = model.predict(test_X)


mse = mean_squared_error(test_y, predictions)
r2 = r2_score(test_y, predictions)

print("Model trained successfully!")
print("Mean Squared Error (MSE):", mse)
print("R-squared (R2) Score:", r2)

print("Model Coefficients:", model.coef_)
print("Model Intercept:", model.intercept_)
```

```python
import pickle

with open("linear_regression_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model saved successfully as 'model.pkl'")
```
